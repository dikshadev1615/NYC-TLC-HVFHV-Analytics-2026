## Data Quality Findings & Cleaning Decisions

### Duplicate Records
- No exact duplicate rows were found.
- **Action:** Retain all records.

### Missing Values
- `originating_base_num` has 5,676,843 missing values.
- 99.7% of HV0005 records have this field missing, while HV0003 has complete coverage.
- **Action:** Retain missing values; no imputation was performed because the missingness is systematic.

### Negative Financial Values
- `base_passenger_fare`: 29,965 negative values.
- `driver_pay`: 56 negative values.
- Other financial fields contain no negative values.
- **Action:** Retain these records and create flags rather than removing them. They may represent financial adjustments or other source-system records and should not be assumed invalid without further documentation.

### Zero-Distance Trips
- 2,782 trips have `trip_miles = 0`.
- 737 trips have both `trip_miles = 0` and `trip_time = 0`.
- These records can contain positive fares and driver pay.
- **Action:** Retain them.

### Extreme Speed / Distance Anomalies
- 13 trips have an implied average speed above 100 mph.
- These records contain internally implausible distance-duration combinations.
- **Action:** Retain the original records but flag them as speed/distance anomalies. They can be excluded from analyses where distance or speed is relevant.

### Timestamp Validation
- No records have `pickup_datetime > dropoff_datetime`.
- `trip_time` is highly consistent with the difference between pickup and dropoff timestamps.
- **Action:** Retain the timestamp and trip-duration fields.

### Long Trips
- 255 trips have `trip_time > 4 hours`.
- Several are plausible long-distance trips.
- **Action:** Do not remove based solely on trip duration.

### Raw Data
- The original January Parquet file remains unchanged.
- Cleaning and derived fields will be created in the `processed data` folder.

```python
import numpy as np

# Create a working copy of the raw data
processed_df = df.copy()

# --------------------------------------------------
# 1. Trip duration in minutes
# --------------------------------------------------
processed_df["trip_duration_minutes"] = (
    processed_df["trip_time"] / 60
)

# --------------------------------------------------
# 2. Average speed in mph
# --------------------------------------------------
processed_df["avg_speed_mph"] = np.where(
    processed_df["trip_time"] > 0,
    processed_df["trip_miles"] / (processed_df["trip_time"] / 3600),
    np.nan
)

# --------------------------------------------------
# 3. Data quality flags
# --------------------------------------------------

# Negative financial values
processed_df["negative_fare_flag"] = (
    processed_df["base_passenger_fare"] < 0
)

processed_df["negative_driver_pay_flag"] = (
    processed_df["driver_pay"] < 0
)

# Zero-distance trips
processed_df["zero_distance_flag"] = (
    processed_df["trip_miles"] == 0
)

# Zero-distance AND zero-duration trips
processed_df["zero_distance_zero_time_flag"] = (
    (processed_df["trip_miles"] == 0) &
    (processed_df["trip_time"] == 0)
)

# Very high implied speed
processed_df["high_speed_flag"] = (
    processed_df["avg_speed_mph"] > 100
)

# Very long trips
processed_df["long_trip_flag"] = (
    processed_df["trip_time"] > 14400
)

# Missing originating base
processed_df["missing_originating_base_flag"] = (
    processed_df["originating_base_num"].isna()
)

# --------------------------------------------------
# 4. Timestamp consistency flag
# --------------------------------------------------
processed_df["invalid_timestamp_flag"] = (
    processed_df["pickup_datetime"] > processed_df["dropoff_datetime"]
)

print("Processed DataFrame created successfully.")
print("Rows:", processed_df.shape[0])
print("Columns:", processed_df.shape[1])
```

Analyze daily trip volume throughout January 2026 to identify unusually high- and low-demand days.

Analyze trip demand by day of week to identify recurring weekday and weekend demand patterns.

Compare average fare, driver pay, and tips across providers to understand differences in trip-level financial performance.

## 6. Trip Distance Distribution

Analyze trips by distance bands to understand the typical travel range and the concentration of short- and long-distance trips.

```python
# Trip distance distribution

distance_bands = pd.cut(
    cleaned_df["trip_miles"],
    bins=[-0.01, 1, 3, 5, 10, 20, float("inf")],
    labels=[
        "0–1 miles",
        "1–3 miles",
        "3–5 miles",
        "5–10 miles",
        "10–20 miles",
        "20+ miles"
    ]
)

distance_distribution = (
    distance_bands
    .value_counts()
    .sort_index()
    .reset_index()
)

distance_distribution.columns = ["distance_band", "trip_count"]

distance_distribution["share_pct"] = (
    distance_distribution["trip_count"]
    / distance_distribution["trip_count"].sum()
    * 100
)

distance_distribution
```

## 7. Trip Duration Distribution

Analyze trips by duration bands to understand typical trip lengths and the concentration of short- and long-duration rides.

```python
# Trip duration distribution

duration_bands = pd.cut(
    cleaned_df["trip_time"] / 60,
    bins=[0, 5, 10, 20, 30, 60, float("inf")],
    labels=[
        "0–5 min",
        "5–10 min",
        "10–20 min",
        "20–30 min",
        "30–60 min",
        "60+ min"
    ],
    include_lowest=True
)

duration_distribution = (
    duration_bands
    .value_counts()
    .sort_index()
    .reset_index()
)

duration_distribution.columns = ["duration_band", "trip_count"]

duration_distribution["share_pct"] = (
    duration_distribution["trip_count"]
    / duration_distribution["trip_count"].sum()
    * 100
)

duration_distribution
```

## 11. Shared Ride Adoption

Analyze shared-ride requests and successful shared matches to understand the adoption of pooled trips within the HVFHV network.
